## 9章コラム NumPy配列に対する集計関数の操作

```python
# データ定義
import numpy as np

x = np.array([[1,2,3],[4,5,6]])
print(x)

# 3通りのmax関数
y = x.max(axis=0)
print(y)

z = x.max(axis=1)
print(z)

w = x.max()
print(w)
```

```python
# keepdimsオプションの利用

y2 = x.max(axis=0, keepdims=True)
y3 = x - y2
print(y2)
print(y3)

z2 = x.max(axis=1, keepdims=True)
z3 = x - z2
print(z2)
print(z3)
```
